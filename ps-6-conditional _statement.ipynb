```python
try:
    num =eval(input("enter a number:"))
    if num%2==0:
        print("the {} is even number".format(num))
    else:
        print("the {} is odd number".format(num))
except Exception as e:
    print(e)
```

```python
# get a random number btw 1 to 200,print you won if number 
# btw 10 to 99,otherwise print you lost
try:
    N=random.randint(1,200)
    if 10<=N<=99:          # N>=10 and N<=99
        print("the num is {} ,you won".format(N))
    else:
        print("the num is {},you lost".format(N))
except Exception as e:
    print(e)
```

**IF-Elif-Else**

```python
try:
    num=eval(input("enter a number"))
    if num>0:
        print("it is postive")
    elif num<0:
        print("it is negtive")
    else:
        print("it is zero")
except Exception as e:
    print(e)
```

```python
# wap enter percentage of marks if per>90 print A or 70-90_B,50-70_c
# otherwise D
try:
    per=eval(input("enter the persentage:"))
    if per>=90:
        print("A grade")
    elif per>=70:
        print("B grade")
    elif per>=50:
        print("C grade")
    else:
        print("D grade")

except Exception as e:
    print(e)
```
