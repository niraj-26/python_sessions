```python
# formate of lambda
# lambda<argument_name>:<output>
```

$One-argument$

```python
def add(x):
    return(x+10)
add(20)  # argu.= X and output= X+10
```

```python
add=lambda x:x+10
add(20)
```

```python
def square(x):
    return(x*x)
square(4) 
```

```python
square=lambda x:x*x
square(5)
```

```python
cube=lambda a:a*a*a
cube(30)
```

$two-argu$

```python
add=lambda a,b:a+b
add(2,6)
```

```python
avg=lambda a,b,c:(a+b+c)/3
avg(25,24,23)
```

```python
# default parameter
avg=lambda a,b,c=30:(a+b+c)/3
avg(25,24,23)
```

```python
avg=lambda a,b,c=30:(a+b+c)/3
avg(25,24)
```

```python
avg=lambda a,b,c=30:round((a+b+c)/3)
avg(25,24)
```

```python
avg=lambda a,b,c=30:round((a+b+c)/3,2)
avg(25,24)                   # give ^ 2 number after point
```

```python
def greater(n1,n2):
    if n1>n2:
        return(n1)
    else:
        return(n2)

greater(100,200)
```

```python
l1=[]            # globle variable 
def greater(n1,n2):
    if n1>n2:
        l1.append(n1)
    else:
        l1.append(n2)

greater(100,200)
l1
```

```python
# lambda <arg1>,<arg2>:<if_output> <if_con> else <else_op>
```

```python
greater=lambda a,b:(a if a>b else b)
greater(8,3)
```

```python
list1=['hyd','mumbai','chennai']
# append method
l2=[]
for i in list1:
    l2.append(i.capitalize())

print(l2)

# list comprehention
[i.capitalize() for i in list1]
```

```python
# lambda <arg>:<op>,<iterator>
lambda i:i.capitalize(),list1
```

```python
# next thing is map input and output
map(lambda i:i.capitalize(),list1)
```

```python
# Store the output
list(map(lambda i:i.capitalize(),list1))
```

- first make a lambda function
- second add your iterator
- map both function and iterator
- save the result in a list
