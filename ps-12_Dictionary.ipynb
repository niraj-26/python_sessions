```python
# wap take 5 random numbers and create a dict with even and odd
d1={}
even=[]
odd=[]
import random
for i in range(5):
    num=random.randint(5,25)
    if num%2==0:
        even.append(num)
        
    else:
        odd.append(num)
d1['even']=even
d1['odd']=odd
print(d1)
```

```python
s1='virat.kohli@blr.com,Rohit.sharma@mumbai.com,kl.rahul@lucknow.com'
l1=s1.split(',')
first_name=[]
second_name=[]
company=[]
d2={}
for i in l1:
    first_name.append(i[0:i.find('.')]  )
    second_name.append(i[i.find('.')+1:i.find('@')])
    company.append(i[i.find('@')+1:i.find('.',i.find('.')+1)])
d2['first_name']=first_name
d2['second_name']=second_name
d2['company']=company
d2
```
