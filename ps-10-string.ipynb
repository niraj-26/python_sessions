**Strip-lstrip-rstrip**

- i want remove the spaces
- if you want to remove spaces both side use strip method
- if you want to remove the spaces only left side then use lstrip: Left strip
- if you want to remove the space only right side then use rstrip: Right strip

```python
str1=' hello how are you '
str2=' hello how are you '
str3="hello how are you "
print(str1.strip())
print(str2.rstrip())
print(str3.lstrip())
```

```python
print(str1.strip())
print(str2.lstrip())
print(str3.rstrip())
```

```python
str1='python.anaconda@nareshit.com'
```

```python
str1.find('.')
s1=str1[:str1.find('.')]
s2=str1[str1.find('.')+1:str1.find('@')]
s3=str1[str1.find('@')+1:str1.find('.',str1.find('.')+1)]
print(s1,s2,s3)
```

```python
str2='omkar.nallagoni@congnizant.com'
print("first name:",str2[:str2.find('.')])
print("second name:",str2[str2.find('.')+1:str2.find('@')])
print("last name:",str2[str2.find('@')+1:str2.find('.',str2.find('.')+1)])
```

```python
str1='omkar.nallagoni@congnizant.com'
n1=str1.find('.')
print("f_name:",str1[:n1])
n2=str1.find('@')
print("s_name:",str1[n1+1:n2])
n3=str1.find('.',str1.find('.')+1)
print("l_name:",str1[n2+1:n3])
```

```python
str1='3.1489'
print(str1[:str1.find('.')])
print(str1[str1.find('.')+1:])
```
