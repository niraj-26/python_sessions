# TypeCasting:

**Type casting (or type conversion) in programming refers to the process of converting a value from one data type to another.**

```python
- Change one data to another data type
- int ---> float
  int ---> string
  int ---> boolean
  int ---> complex

- float ---> int
  float ---> string
  float ---> boolean
  float ---> complex
```

```python
number1 = 100
type(number1)
```

```python
# Case 1: convert int to float
float_number = float(number1)
float_number
```

```python
type(float_number)
```

```python
# case2: int to string
str_number = str(number1)
str_number
```

```python
type(str_number)
```

```python
# case3: int to bool
bool_number = bool(number1)
bool_number
```

```python
type(bool_number)
```

```python
bool(0)
```
