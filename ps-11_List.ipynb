**append vs extend vs concat**

- append will add the element at last, that result also save in a same list
  
      list1=['A','B','C']
      list2=[1,2,3]
      list1.extend(list2)=[1,2,3,['A','B']]
- Extend will update the list by adding new elements, the result also save in a same list
  
      list1=['A','B','C']
      list2=[1,2,3]
      list1.extend(list2)=[1,2,3,'A','B']
- Concat will add two list,but the result will not update in a list
  
      list1=['A','B','C']
      list2=[1,2,3]
      list1.extend(list2)=[1,2,3,'A','B']
